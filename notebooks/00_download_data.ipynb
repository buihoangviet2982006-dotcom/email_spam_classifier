# 00. Tải Dữ liệu Gốc từ Kaggle

Vì file dữ liệu gốc `combined_data.csv` (~133 MB) vượt quá giới hạn dung lượng file của GitHub (100 MB), notebook này tự động tải bộ dữ liệu từ Kaggle và lưu vào `data/raw/combined_data.csv`.

- Nguồn dữ liệu: [`purusinghvi/email-spam-classification-dataset`](https://www.kaggle.com/datasets/purusinghvi/email-spam-classification-dataset)
- Yêu cầu: Đã cài `kagglehub` (`pip install -r requirements.txt`). Nếu máy yêu cầu xác thực Kaggle, cấu hình `~/.kaggle/kaggle.json` hoặc biến môi trường `KAGGLE_USERNAME` / `KAGGLE_KEY`.

```python
import shutil
from pathlib import Path

import kagglehub
import pandas as pd

# Tự động tìm thư mục gốc dự án
cwd = Path.cwd().resolve()
PROJECT_DIR = next(
    (p for p in [cwd, *cwd.parents] if (p / "src").is_dir() and (p / "notebooks").is_dir()),
    cwd,
)

raw_dir = PROJECT_DIR / "data" / "raw"
raw_dir.mkdir(parents=True, exist_ok=True)
target_csv = raw_dir / "combined_data.csv"

if target_csv.is_file():
    print(f"Dữ liệu đã tồn tại tại: {target_csv}")
else:
    print("Đang tải bộ dữ liệu từ Kaggle...")
    download_dir = Path(kagglehub.dataset_download("purusinghvi/email-spam-classification-dataset"))
    source_csv = download_dir / "combined_data.csv"
    if not source_csv.is_file():
        raise FileNotFoundError(f"Không tìm thấy combined_data.csv trong {download_dir}")
    shutil.copy2(source_csv, target_csv)
    print(f"Đã lưu dữ liệu gốc tại: {target_csv}")

size_mb = target_csv.stat().st_size / (1024 * 1024)
df = pd.read_csv(target_csv)
print(f"Dung lượng: {size_mb:.2f} MB | Kích thước: {df.shape[0]:,} dòng, {df.shape[1]} cột")
print(df.head(3))
```
